Import lib

```python
import pandas as pd
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# display options
pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 200)
sns.set(style='whitegrid', rc={'figure.figsize':(8,5)})

BASE_DIR = pathlib.Path().resolve().parent.parent
target_column = 'isFraud'
```

Load train and identity files

```python
# Load 
transaction_file_name = BASE_DIR / "data/raw/ieee-fraud-detection/train_transaction.csv"
identity_file_name = BASE_DIR / "data/raw/ieee-fraud-detection/train_identity.csv"

df_transaction = pd.read_csv(transaction_file_name)
df_identity = pd.read_csv(identity_file_name)
```

```python
# Merge on TransactionID
df_both = df_transaction.merge(df_identity, how='left', on='TransactionID') 
```

Check general information

```python
# General
df = df_both

print("=== SHAPE ===")
print(df.shape)
print("\n=== DTYPE COUNTS ===")
print(df.dtypes.value_counts())
print("\n=== HEAD ===")
display(df.head())
print("\n=== INFO() ===")
display(df.info())
print("\n=== DESCRIBE (numeric) ===")
display(df.describe().T)
print("\n=== SAMPLE NULLS (top 20 cols with most nulls) ===")
nulls = df.isnull().sum().sort_values(ascending=False).head(20)
display(nulls.to_frame(name='null_count'))
```

NOW: Drop row and column with > 50% nan

LATER: Analyse if missing values is connect with target

```python
# Check target 
print("Columns:", df.columns.tolist()[:200])
print("Target dtype =",df[target_column].dtype)
print(df[target_column].value_counts(dropna=False))
print(df[target_column].value_counts(normalize=True, dropna=False).round(3))
```

```python
# Check nan
nulls = df.isnull().sum()
nulls = nulls[nulls>0].sort_values(ascending=False)
nulls_pct = (nulls / len(df) * 100).round(3)
missing_df = pd.DataFrame({'null_count': nulls, 'null_pct': nulls_pct})
high_missing = missing_df[missing_df['null_pct'] > 50]
print(len(high_missing))
```

```python
# === DROP COLUMNS with >50% missing ===
nulls = df.isnull().sum()
nulls_pct = (nulls / len(df) * 100)
high_missing_cols = nulls_pct[nulls_pct > 50].index.tolist()

print(f"Dropping {len(high_missing_cols)} columns with >50% missing:")
print(high_missing_cols)

df_cleaned = df.drop(columns=high_missing_cols)
print(f"\nAfter dropping columns: {df.shape} -> {df_cleaned.shape}")

# === DROP ROWS with >50% missing ===
row_nulls = df_cleaned.isnull().sum(axis=1)
row_nulls_pct = (row_nulls / len(df_cleaned.columns) * 100)
rows_to_drop = row_nulls_pct[row_nulls_pct > 50]

print(f"\nDropping {len(rows_to_drop)} rows with >50% missing ({len(rows_to_drop)/len(df_cleaned)*100:.2f}% of remaining rows)")

df_cleaned = df_cleaned[row_nulls_pct <= 50]

# === FINAL RESULT ===
print(f"\nFinal cleaning summary:")
print(f"Original shape: {df.shape}")
print(f"Final shape: {df_cleaned.shape}")
print(f"Rows removed: {len(df) - len(df_cleaned)}")
print(f"Columns removed: {df.shape[1] - df_cleaned.shape[1]}")

# Save to CSV
df_cleaned.to_csv(BASE_DIR / "data/preprocessed/both_preprocessed.csv", index=False)
print("\nSaved train_transaction_preprocessed.csv")
```

NOTE: Should split first

```python
from sklearn.preprocessing import StandardScaler

# Separate target column BEFORE normalization
y = df_cleaned['isFraud'].copy()
X = df_cleaned.drop('isFraud', axis=1)

print(f"\nFeatures to normalize: {X.shape}")
print(f"Target (will NOT be normalized): {y.shape}")
print(f"Target values: {y.unique()}")  # Should be [0, 1]

# Check for any remaining nulls in features
if X.isnull().sum().sum() > 0:
    print("\n⚠️  WARNING: Found nulls in features before normalization!")
    print(X.isnull().sum()[X.isnull().sum() > 0])
    print("\nFilling remaining nulls with 0...")
    X = X.fillna(0)

# Initialize and fit scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame
X_scaled = pd.DataFrame(X_scaled, columns=X.columns, index=X.index)

# Verify normalization
print(f"\nNormalized features:")
print(f"  Mean (should be ~0): {X_scaled.mean().mean():.6f}")
print(f"  Std (should be ~1): {X_scaled.std().mean():.6f}")
print("✓ Normalization complete")
```

```python
# === STEP 4: Combine with ORIGINAL Target and Save ===
print("\n" + "="*80)
print("STEP 4: SAVING PREPROCESSED DATA")
print("="*80)

# Combine normalized features with ORIGINAL target
df_final = pd.concat([X_scaled, y.reset_index(drop=True)], axis=1)

# Verify target is unchanged
print(f"\nTarget column check:")
print(f"  Unique values: {df_final['isFraud'].unique()}")  # Should be [0, 1]
print(f"  Mean (fraud rate): {df_final['isFraud'].mean():.4f}")
print(f"  Min: {df_final['isFraud'].min()}, Max: {df_final['isFraud'].max()}")

# Check for any nulls in final data
final_null_check = df_final.isnull().sum().sum()
print(f"\nNull values in final dataset: {final_null_check}")
if final_null_check > 0:
    print("⚠️  WARNING: Final dataset still has nulls!")
    print(df_final.isnull().sum()[df_final.isnull().sum() > 0])

print(f"\nFinal shape: {df_final.shape}")

# Save to CSV
preprocessed_dir = BASE_DIR / "data/preprocessed"
preprocessed_dir.mkdir(parents=True, exist_ok=True)

output_path = preprocessed_dir / "both_preprocessed_normalized.csv"
df_final.to_csv(output_path, index=False)
print(f"✓ Saved to: {output_path}")

# Save the scaler
import joblib
scaler_path = preprocessed_dir / "scaler.pkl"
joblib.dump(scaler, scaler_path)
print(f"✓ Saved scaler to: {scaler_path}")

# === FINAL SUMMARY ===
print("\n" + "="*80)
print("PREPROCESSING COMPLETE!")
print("="*80)
print(f"Final shape: {df_final.shape}")
print(f"Features (normalized): {X_scaled.shape[1]}")
print(f"Target (NOT normalized): isFraud ∈ {df_final['isFraud'].unique()}")
print(f"Samples: {len(df_final)}")
print(f"Fraud rate: {df_final['isFraud'].mean():.4f}")
print(f"No nulls: {df_final.isnull().sum().sum() == 0}")
```
